```python
import numpy as np
import pandas as pd
arr = [10, 15, 22, 35, 40, 55]

print(" List:  ", arr)
print("3rd Element:", arr[2])       
print("First Three:", arr[:3])      
print("Last Element:", arr[-1])  


matrix_2d = np.array([[10,20,30] ,[40,50,60] ,[70, 80, 90]])

print("Full 2D Matrix:\n", matrix_2d)

print("Specific Element [Row 1, Col 2]:", matrix_2d[1, 2]) 

print(" elements to print:\n", matrix_2d[0:2, 1:3]) 



student_data = {
    "Student_Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Branch": ["CSE", "ECE", "CSE", "MECH"],
    "Math_Marks":[20,40,60,80]
}
df = pd.DataFrame(student_data)
print("Initial Pandas DataFrame:\n", df)

print("\nPandas .loc Slicing (Rows 0-2, specific columns):")
print(df.loc[0:2, ["Student_Name", "Math_Marks"]])

print("\nPandas .iloc Slicing (Row 1 to 2, Column 2 onwards):")
print(df.iloc[1:3, 2:])
```
